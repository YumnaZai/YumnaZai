#Handling missing values

```python
# 1) Confirm TotalCharges is numeric and count missing values
Telecommuniction_df['TotalCharges'] = pd.to_numeric(Telecommuniction_df['TotalCharges'], errors='coerce')
missing_count = Telecommuniction_df['TotalCharges'].isna().sum()
total = len(Telecommuniction_df)
print(f"Missing TotalCharges: {missing_count} / {total} ({missing_count/total*100:.6f}%)")


# 2) Inspect the missing rows
missing_rows = Telecommuniction_df[Telecommuniction_df['TotalCharges'].isna()]
print(missing_rows[['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']])

# 3) Check if tenure==0 for these rows (this is the key check)
print(missing_rows['tenure'].value_counts(dropna=False))
```

When the records corresponding to the missing values are inspected, it can be observed that the tenure == 0 where Total charges is indicated as Nan.


Therefore, if those rows have tenure == 0: set TotalCharges = 0 for them

Method of handling missing values -> Imputation
Why imputataion-
- No deleteion -> Dropping rows loses actual customer records. Even 11 rows can represent a meaningful group
- Imputation preserves the original distribution
- In the context of the tele communication domain it is known that when tunure is zero the total charges should be zero, theerefore a resonable value like 0 can be imputed for toatal charges instead of ,mean/ median , which would be wrong
- Reduces noise caused by NaNs and prevents the introduction of spurious patterns that naive deletion or random filling might create.


```python

# 4) Create a missing-flag
# why - models can learn whether “was missing” affects outcomes
Telecommuniction_df['TotalCharges_missing_flag'] =Telecommuniction_df['TotalCharges'].isna().astype(int)

# 5) Conditional imputation: if tenure == 0 -> TotalCharges = 0
Telecommuniction_df.loc[(Telecommuniction_df['TotalCharges'].isna()) & (Telecommuniction_df['tenure'] == 0), 'TotalCharges'] = 0
```

```python
#Check if there are any missing values remaining
remaining_nans = Telecommuniction_df['TotalCharges'].isna().sum()
print("Remaining NaNs after conditional imputation:", remaining_nans)
```

```python
#IQR Method
# Extracting the numeric features
data = Telecommuniction_df.drop(['customerID', 'gender', 'SeniorCitizen', 'Partner', 'Dependents',
      'PhoneService', 'MultipleLines', 'InternetService',
       'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport',
       'StreamingTV', 'StreamingMovies', 'Contract', 'PaperlessBilling',
       'PaymentMethod', 'Churn'], axis = 1)

Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

outliers_iqr = ((data < (Q1 - 1.5 * IQR)) | (data > (Q3 + 1.5 * IQR)))
print(outliers_iqr.sum())
```
